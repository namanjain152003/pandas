```python
s=pd.Series([1,2,3,4,5])
print(s, type(s))

# indexing
print(s[1])
print(s[3])
print(s[2])
```

```python
s=pd.Series([24,25,26,27], index=["naman","harshit","rajat","rahul"])
print(s)

print(s["naman"])
print(s["rahul"])

print(s.iloc[0])  # used to access the elements by index if customized labels are used
```

```python
# properties of the 

# homogeneous means we can store any type of data but it will make all the data as same data type

# Vectorized Operations
s1 = pd.Series([1, 2, 3])
s2 = pd.Series([4, 5, 6])
print(s1 + s2)

# handle missing values with NaN

# Mutable Values but immutable size
s = pd.Series([1, 2, 3, 4, 5])
s[0] = 100
print(s)

changed_s = s.drop(1)
print(changed_s)
print(s)
```
