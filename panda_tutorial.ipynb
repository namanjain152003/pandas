```python
# cleaning of the data

df=pd.read_csv("raw_data.csv")
df

# Handle Missing values
df.isnull() # True for null values (NaN)

df.isnull().sum() # counts missing vals per column

df.dropna() # drops rows with missing values

df.dropna(axis = 1) # drops cols with missing values

df.fillna(0) # fills NaN with 0

df["age"] = df["age"].fillna(df["age"].mean()) # fills column with mean

df.ffill() # forward fill

df.bfill() # backward fill
```

```python
# handle duplicates values
df.duplicated() # True for duplicate rows
df.duplicated("country") # True for duplicate rows in particular col
df.duplicated(["country", "gender"])

df2=df.copy()
df.drop_duplicates(inplace=True) # drops duplicate rows
df
```

```python
df["gender"].str.lower()
# .str.lower() - Convert to lowercase

df["gender"].str.upper()
# .str.upper() - Convert to uppercase

df["gender"].str.capitalize()
# .str.capitalize() - Capitalize strings

df["name"].str.strip()
#.str.strip() - Remove leading/trailing spaces

df["name"].str.split(" ")
# .str.split(" ")- Split into parts based on a separator

df["country"].str.contains("US")
df["country"].str.contains("india" , case=False) # case=False will not take care of the case sensitivity
# .str.contains()  - Check if a value exists in string or not
```

```python
# DATA TRANSFORMATION

df2= df.copy()

df2["tax"] =df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%")

df2
```

```python
gender_map = {"Male" : "M", "Female" : "F", "Unknown" :"U"}
df2["gender"] = df2["gender"].map(gender_map)
df2
```

```python
df2=df2.assign(new_income =df2["income"] *1.1)
```

```python
df2["country"] = df2["country"].replace("USA", "United States")
df2
```

```python
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]
df2.rename(columns={"Income":"Salary"})
```
